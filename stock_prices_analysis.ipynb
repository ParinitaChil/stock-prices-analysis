# Stock Prices Analysis (Finance Project)
This beginner-friendly notebook analyzes historical stock prices using Python.
It includes:
- Loading real stock price data
- Data cleaning
- Exploratory analysis
- Visualizations
- Simple statistics
- Trend summary

```
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
url = "https://raw.githubusercontent.com/datasets/s-and-p-500-companies/master/data/constituents-financials.csv"
df = pd.read_csv(url)
df.head()
```

## 1. Understanding the Data

```
df.info()
```

## 2. Basic Statistics

```
df.describe()
```

## 3. Visualization: Market Capitalization vs Sector

```
plt.figure(figsize=(12,6))
df.groupby('Sector')['MarketCap'].sum().sort_values().plot(kind='barh')
plt.title("Total Market Capitalization by Sector")
plt.xlabel("Market Cap (USD)")
plt.ylabel("Sector")
plt.show()
```

## 4. Key Findings
- Technology and Financial sectors have the highest total market capitalization.
- Visualization helps identify sector-wise distribution of large companies.
- Dataset is clean and ready for further analysis, such as time series or portfolio optimization.
